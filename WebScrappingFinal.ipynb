```python
!pip install -q python-jobspy pandas

import pandas as pd
from jobspy import scrape_jobs
from google.colab import files
```

```python
print("Mega Scraping is starting, it may take few minutes..")
```

```python
# Step 2: Multiple roles
job_roles = [
    "Data Analyst", "Business Analyst", "Data Scientist", "Software developer",
    "management", "Sales manager", "Senior accountant", "AI/ML engineer",
    "Data Engineer", "Full Stack Developer", "Frontend Developer", "Backend Developer",
    "DevOps Engineer", "Cloud Architect", "Product Manager", "Project Manager",
    "Marketing Manager", "Financial Analyst", "HR Manager", "Operations Manager",
    "UI/UX Designer", "Quality Assurance Engineer", "Cyber Security Analyst",
    "Systems Administrator", "Network Engineer", "Database Administrator",
    "Technical Support Engineer", "Content Writer", "Graphic Designer",
    "Business Development Executive"
]
```

```python
all_jobs = []

for role in job_roles:
    print(f"'{role}' job scraping is started...")
    try:
        jobs = scrape_jobs(
            site_name=["indeed", "linkedin"],
            search_term=role,
            location="India",
            results_wanted=100,  # for each role Har role 100 jobs
            country_indeed='India'
        )
        all_jobs.append(jobs)
    except Exception as e:
        print(f"Error in {role}: {e}")
```

```python
# Step 3: merging all the data
df_jobs = pd.concat(all_jobs, ignore_index=True)
print(f"Total raw jobs without filtering: {len(df_jobs)}")

# important coloumns:--
selected_columns = [
    'title',
    'company',
    'location',
    'job_type',
    'min_amount',
    'max_amount',
    'date_posted',
    'description',
    'job_url',
    'site'
]
final_columns = [col for col in selected_columns if col in df_jobs.columns]
df_jobs = df_jobs[final_columns].copy()
```

```python
# Data Cleaning
if 'description' in df_jobs.columns:
    df_jobs['description'] = df_jobs['description'].astype(str).str.replace('\n', ' ', regex=False).str.replace('\r', ' ', regex=False)
if 'title' in df_jobs.columns:
    df_jobs['title'] = df_jobs['title'].astype(str).str.replace('\n', ' ', regex=False).str.replace('\r', ' ', regex=False)

# -------------- STRICT SKIP --------------
df_jobs = df_jobs.dropna()

# remove duplicate jobs
df_jobs = df_jobs.drop_duplicates(subset=['title', 'company', 'location'])
# -----------------------------------------

# Step 4: Final Perfect Data Save
csv_filename = "CodeAlpha_JobData_Mega.csv"
df_jobs.to_csv(csv_filename, index=False, encoding="utf-8-sig")

print(f"wihtout missing values data is saved: {len(df_jobs)}")

files.download(csv_filename)
```
